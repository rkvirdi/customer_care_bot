RAGAS (Retrieval-Augmented Generation Assessment Suite) is an open-source framework designed to evaluate and measure the performance of Retrieval-Augmented Generation (RAG) systems.  

RAGAS provides objective metrics to assess the different components of a RAG pipeline—the retrieval (finding relevant documents) and the generation (creating the answer) 

Key RAGAS Metrics for Evaluation
RAGAS focuses on various aspects of a RAG system's performance. The main component-level metrics include:   

For the Generative Component (The Answer)
Faithfulness: Measures the factual consistency of the generated answer against the given retrieved context. A high score means the answer is supported by the facts in the context, helping to mitigate LLM hallucinations.   

Answer Relevancy: Assesses how pertinent the generated answer is to the original question. A low score might result from an answer that is incomplete, contains redundant information, or drifts from the user's intent.   

For the Retrieval Component (The Context)
Context Precision: Evaluates the signal-to-noise ratio of the retrieved context. It measures whether the relevant information needed to answer the question is ranked highly or is buried among irrelevant documents.   

Context Recall: Measures if all the relevant information required to generate a good answer (often based on a human-provided or synthetically generated "ground truth") was successfully retrieved by the system. 

| Metric                | Meaning                                                                    |
| --------------------- | -------------------------------------------------------------------------- |
| **Faithfulness**      | Checks if the answer is supported by retrieved context (no hallucination). |
| **Answer Relevancy**  | How relevant the answer is to the user query.                              |
| **Context Precision** | Measures how much of the retrieved context is actually useful.             |
| **Context Recall**    | Measures how much of the relevant information was retrieved.               |

```python
import pandas as pd
from datasets import Dataset
from ragas import evaluate
from ragas.metrics import faithfulness, answer_relevancy, context_precision, context_recall
```

```python
data = {
    "question": [
        "What safety features are included in Model S?",
        "How do I activate Autopilot in Model X?"
    ],
    "contexts": [
        [
            "Model S includes Lane Assist, Collision Avoidance, and Speed Assist.",
            "Autopilot also offers Traffic-Aware Cruise Control."
        ],
        [
            "To activate Autopilot, press down twice on the right stalk.",
            "Ensure lane markings are visible and conditions are safe."
        ]
    ],
    "ground_truth": [
        "Model S includes Lane Assist, Collision Avoidance, and Speed Assist.",
        "Autopilot activates by pressing down twice on the right stalk."
    ],
    "answer": [
        "Model S comes with lane assist, collision avoidance, and speed assist.",
        "Press down twice on the right stalk to turn on Autopilot."
    ]
}

dataset = Dataset.from_pandas(pd.DataFrame(data))
dataset
```

```python
results = evaluate(
    dataset=dataset,
    metrics=[faithfulness, answer_relevancy, context_precision, context_recall]
)
results
```
